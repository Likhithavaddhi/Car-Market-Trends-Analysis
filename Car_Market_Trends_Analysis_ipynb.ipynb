```python
print("========== CAR MARKET SUMMARY ==========")

print("Total Cars:", len(df))
print("Average Selling Price:", round(df['Selling_Price'].mean(), 2))
print("Average Present Price:", round(df['Present_Price'].mean(), 2))
print("Average KMs Driven:", round(df['Kms_Driven'].mean(), 2))
print("Average Car Age:", round(df['Car_Age'].mean(), 2))

print("\nMost Common Fuel Type:")
print(df['Fuel_Type'].value_counts().idxmax())

print("\nMost Common Transmission:")
print(df['Transmission'].value_counts().idxmax())

print("\nMost Common Seller Type:")
print(df['Seller_Type'].value_counts().idxmax())
```

```python
fuel_avg = df.groupby('Fuel_Type')['Selling_Price'].mean()
trans_avg = df.groupby('Transmission')['Selling_Price'].mean()
seller_avg = df.groupby('Seller_Type')['Selling_Price'].mean()

print("Highest Average Selling Price by Fuel Type:")
print(fuel_avg.idxmax(), "=", round(fuel_avg.max(), 2))

print("\nHighest Average Selling Price by Transmission:")
print(trans_avg.idxmax(), "=", round(trans_avg.max(), 2))

print("\nHighest Average Selling Price by Seller Type:")
print(seller_avg.idxmax(), "=", round(seller_avg.max(), 2))
```

```python
fig, axes = plt.subplots(1, 3, figsize=(16, 5))

df.groupby('Fuel_Type')['Selling_Price'].mean().plot(
    kind='bar', ax=axes[0]
)
axes[0].set_title('Average Price by Fuel Type')
axes[0].set_xlabel('Fuel Type')
axes[0].set_ylabel('Selling Price')

df.groupby('Transmission')['Selling_Price'].mean().plot(
    kind='bar', ax=axes[1]
)
axes[1].set_title('Average Price by Transmission')
axes[1].set_xlabel('Transmission')
axes[1].set_ylabel('Selling Price')

df.groupby('Seller_Type')['Selling_Price'].mean().plot(
    kind='bar', ax=axes[2]
)
axes[2].set_title('Average Price by Seller Type')
axes[2].set_xlabel('Seller Type')
axes[2].set_ylabel('Selling Price')

plt.tight_layout()
plt.show()
```

```python
# Final statistical insights

print("========== FINAL KEY INSIGHTS ==========\n")

# 1. Present Price correlation
corr = df['Present_Price'].corr(df['Selling_Price'])
print(f"1. Correlation between Present Price and Selling Price: {corr:.2f}")

# 2. Car Age correlation
age_corr = df['Car_Age'].corr(df['Selling_Price'])
print(f"2. Correlation between Car Age and Selling Price: {age_corr:.2f}")

# 3. KMs Driven correlation
kms_corr = df['Kms_Driven'].corr(df['Selling_Price'])
print(f"3. Correlation between KMs Driven and Selling Price: {kms_corr:.2f}")

# 4. Fuel type
fuel_avg = df.groupby('Fuel_Type')['Selling_Price'].mean()
print(f"\n4. Fuel type with highest average selling price: {fuel_avg.idxmax()}")

# 5. Transmission
trans_avg = df.groupby('Transmission')['Selling_Price'].mean()
print(f"5. Transmission with highest average selling price: {trans_avg.idxmax()}")

# 6. Seller type
seller_avg = df.groupby('Seller_Type')['Selling_Price'].mean()
print(f"6. Seller type with highest average selling price: {seller_avg.idxmax()}")

# 7. Most expensive car
expensive = df.loc[df['Selling_Price'].idxmax(), 'Car_Name']
print(f"7. Most expensive car in the dataset: {expensive}")
```

## 📌 Project Overview

This project analyzes used-car market data from CarDekho to identify important
patterns and factors affecting the selling price of vehicles.

The analysis was performed using Python, Pandas, Matplotlib, and Seaborn in
Google Colab.

The project explores vehicle characteristics such as present price,
selling price, manufacturing year, kilometers driven, fuel type,
seller type, transmission, and previous ownership.

---

## 🎯 Objectives

- Understand the used-car dataset.
- Clean and prepare the data for analysis.
- Analyze the distribution of car prices.

- Compare manual and automatic vehicles.
- Compare dealer and individual sellers.
- Identify the top-selling-price vehicles.
- Generate meaningful business insights.
