```python
# Car Market Trends Analysis Using CarDekho Data

## 📌 Project Overview

This project analyzes used-car market data from CarDekho to identify trends,
patterns, and factors that influence the selling price of vehicles.

The project uses Python-based data analysis techniques to clean, explore,
visualize, and interpret the dataset.

## 🎯 Objectives

- Analyze used-car market data.
- Clean and prepare the dataset.
- Understand car price distributions.
- Analyze the relationship between Present Price and Selling Price.
- Study the effect of car age and kilometers driven.
- Compare cars based on fuel type, transmission, and seller type.
- Identify the top 10 cars by selling price.
- Generate meaningful business insights.

## 📊 Dataset

The dataset contains information about used cars.

| Column | Description |
|---|---|
| Car_Name | Name of the car |
| Year | Manufacturing year |
| Selling_Price | Selling price of the car |
| Present_Price | Present price of the car |
| Kms_Driven | Kilometers driven |
| Fuel_Type | Fuel type |
| Seller_Type | Dealer or Individual |
| Transmission | Manual or Automatic |
| Owner | Number of previous owners |

## 🛠️ Technologies Used

- Python
- Pandas
- NumPy
- Matplotlib
- Seaborn
- Google Colab

## 🔄 Data Analysis Process

1. Data loading
2. Data exploration
3. Data cleaning
4. Missing-value and duplicate checking
5. Feature engineering
6. Exploratory Data Analysis (EDA)
7. Data visualization
8. Correlation analysis
9. Business insights
10. Conclusion

## 🧹 Data Cleaning

The dataset was checked for:

- Missing values
- Duplicate records
- Incorrect data types
- Data consistency

Two additional features were created:

- `Car_Age`
- `Price_Drop`

## 📈 Analysis Performed

The project includes analysis of:

- Fuel type distribution
- Transmission distribution
- Seller type distribution
- Selling price distribution
- Present Price vs Selling Price
- Car Age vs Selling Price
- Kilometers Driven vs Selling Price
- Year-wise average selling price
- Top 10 cars by selling price
- Correlation between numerical variables

## 🔎 Key Findings

- Present Price has a strong positive relationship with Selling Price.
- The correlation between Present Price and Selling Price is approximately 0.88.
- Car Age has a negative relationship with Selling Price.
- Previous ownership has a weak relationship with Selling Price.
- Different fuel types, transmission types, and seller types show differences
  in average selling prices.

## 💼 Business Recommendations

- Present Price should be considered when estimating resale value.
- Vehicle age should be considered when pricing used cars.
- Mileage should be evaluated along with other vehicle characteristics.
- Historical data can help dealers make better pricing decisions.
- Buyers can compare vehicle characteristics before purchasing a used car.

## 🏁 Conclusion

This project demonstrates how Python and data analysis techniques can be used
to transform raw used-car data into meaningful insights.

The analysis provides useful information about used-car pricing and market
trends and demonstrates practical skills in data cleaning, exploratory data
analysis, visualization, and business interpretation.

## 👩‍💻 Author

**Likhitha Vaddhi**

B.Tech – Computer Science and Data Science
```
